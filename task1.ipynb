```python
import matplotlib.pyplot as plt
avg_sepal = data.groupby("Species")["SepalLengthCm"].mean()

plt.figure(figsize=(6,4))
avg_sepal.plot(kind="bar")
plt.title("Average Sepal Length by Species")
plt.xlabel("Species")
plt.ylabel("Average Sepal Length (cm)")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.hist(data["SepalLengthCm"], bins=10)
plt.title("Distribution of Sepal Length")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(6,4))

for species in data["Species"].unique():
    subset = data[data["Species"] == species]
    plt.scatter(
        subset["SepalLengthCm"],
        subset["PetalLengthCm"],
        label=species
    )

plt.title("Sepal Length vs Petal Length")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Petal Length (cm)")
plt.legend()
plt.show()
```

```python
comparison = data.groupby("Species")[[
    "SepalLengthCm",
    "SepalWidthCm",
    "PetalLengthCm",
    "PetalWidthCm"
]].mean()

comparison.plot(kind="bar", figsize=(10,5))
plt.title("Average Feature Comparison by Species")
plt.xlabel("Species")
plt.ylabel("Average Value (cm)")
plt.xticks(rotation=0)
plt.legend(title="Features")
plt.show()
```
