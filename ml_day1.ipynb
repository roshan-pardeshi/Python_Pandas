```python

import pandas as pd
import numpy as np

np.random.seed(101)
n = 1000

data = {
    "House_ID": np.arange(1, n + 1).tolist(),

    "Area_sqft": np.random.randint(500, 4001, n).tolist(),

    "Bedrooms": np.random.randint(1, 7, n).tolist(),

    "Bathrooms": np.random.randint(1, 5, n).tolist(),

    "Age_years": np.random.randint(0, 51, n).tolist(),

    "Distance_to_City_km": np.round(
        np.random.uniform(1, 40, n), 1
    ).tolist(),

    "Parking_Spaces": np.random.randint(0, 5, n).tolist(),

    "Floor_Number": np.random.randint(0, 21, n).tolist(),

    "Nearby_Schools": np.random.randint(0, 11, n).tolist(),

    "Crime_Rate": np.round(
        np.random.uniform(1, 10, n), 1
    ).tolist(),

    "Property_Tax": np.random.randint(5000, 50001, n).tolist()
}

# Generate synthetic house prices
noise = np.random.normal(0, 500000, n)

price = (
    800000
    + np.array(data["Area_sqft"]) * 2500
    + np.array(data["Bedrooms"]) * 300000
    + np.array(data["Bathrooms"]) * 200000
    - np.array(data["Age_years"]) * 25000
    - np.array(data["Distance_to_City_km"]) * 20000
    + np.array(data["Parking_Spaces"]) * 150000
    + np.array(data["Floor_Number"]) * 10000
    + np.array(data["Nearby_Schools"]) * 50000
    - np.array(data["Crime_Rate"]) * 40000
    - np.array(data["Property_Tax"]) * 2
    + noise
)

data["Price_INR"] = np.maximum(
    price, 500000
).round(0).astype(int).tolist()

df = pd.DataFrame(data)

print(df.head(10))
print("Dataset shape:", df.shape)
print(df.info())
```
