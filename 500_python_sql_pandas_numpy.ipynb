```python
count = 0 
def prime_number(n):
    factor = 0
    num = n
    count = 0 
    for i in range(2,num,1):
        if num%2==0:
            factor+=1
    if factor==0:
        print("prime number----",num)
        count+=1
    
        
    
       
for i in range(1,101,1):
    prime_number(i)
print(count)
```
