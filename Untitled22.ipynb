```python
def sonu(batsman):
    mark = delivery['batsman_runs']==6
    new= delivery[mark]
    vk = delivery[delivery['batsman']==batsman]
    return vk.groupby('batting_team')['batsman_runs','batting_team'].sum().sort_values(ascending=False)
    new()
```
