```python
class arithmatic:
    def add(self,a,b):
        print(a+b)

class substraction:
    def sub(self,c,d):
        print(c+d)

class allcalculation(arithmatic,substraction):
    def all(self):
        print("successfully")

a = allcalculation()
a.add(10,20)
a.sub(30,40)
    
```
