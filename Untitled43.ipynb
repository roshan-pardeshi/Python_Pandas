```python
df = pd.DataFrame({
    "Salary": [20000, 50000, 80000]
})

def category(s):
    if s < 30000:
        return "Low"
    elif s < 70000:
        return "Medium"
    else:
        return "High"

df["Category"] = df["Salary"].apply(category)
print(df)
```
