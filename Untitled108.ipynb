```python

from multipledispatch import dispatch


class A:

    @dispatch(int)
    def operation(self,a):
        print(a**2)

    @dispatch(str,str)
    def operation(self,name,age):
        print(f"the name is a {name} the age is a {age}")

    @dispatch(int,int,int)
    def operation(self,a,b,c):
        print(a+b+c)

a = A()
a.operation(20)
a.operation("roshan","20 above")
a.operation(10,20,30)
```
