```python
def factorials(n):
    fact = 1
    for i in range(1,n+1,1):
        fact*=i
    return fact

num = 14
sum1 = 0
for i in str(num):
    sum1+=factorials(int(i))

if sum1==num:
    print("strong number",sum1)
else:
    print("no strong number")
```

```python
num = 111111

no = num 

rev = 0

while no>0:
    a = no%10
    rev = rev*10+a
    no = no//10

if rev == num:
    print("palindrome number")
else:
    print("not palindrome number")
```

```python
num = 370

a = len(str(num))
cube = 0
no = num

for i in range(1,num+1,1):
    a1 = no%10
    cube = cube + a1**a
    no = no//10
    
if num==cube:
    print("amstrong number")

```

```python
li = [2,7,8,4,2,1]

li2 = []
target = 9 
for i in range(0,len(li),1):
    for j in range(i+1,len(li)):
        if li[i]+li[j]==target:
            li2.append({li[i],li[j]})
            

print(type(li2))
```
